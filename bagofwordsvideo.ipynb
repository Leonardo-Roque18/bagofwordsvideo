```python
import re
import numpy as np
import pandas as pd
```

```python
def meu_bag_of_words(textos):
    # 1. Tokenização: minúsculas + só palavras (sem pontuação)
    tokens_por_frase = []
    for frase in textos:
        tokens = re.findall(r"\w+", frase.lower())
        tokens_por_frase.append(tokens)

    # 2. Vocabulário: palavras únicas, ordenadas
    vocabulario = sorted(set(t for tokens in tokens_por_frase for t in tokens))
    word_to_idx = {palavra: i for i, palavra in enumerate(vocabulario)}

    # 3. Matriz de contagem (lista de listas)
    matriz = []
    for tokens in tokens_por_frase:
        vetor = [0] * len(vocabulario)
        for t in tokens:
            vetor[word_to_idx[t]] += 1
        matriz.append(vetor)

    return matriz, vocabulario
```

```python
frases_teste = [
    "O rato roeu a roupa",
    "A roupa do rei",
    "O Rei roeu a roupa, a roupa do rato!"
]

matriz, vocab = meu_bag_of_words(frases_teste)

print("Vocabulário:", vocab)
print("Matriz:", matriz)
```

```python
print(np.array(matriz).shape)   # (frases, palavras)
pd.DataFrame(matriz, columns=vocab, index=["frase 1", "frase 2", "frase 3"])
```

```python
m, v = meu_bag_of_words(["O rato roeu o rei", "O rei roeu o rato"])
print(v)
print(m)
print("Vetores idênticos?", m[0] == m[1])
```
